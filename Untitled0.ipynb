```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

シークレットにしてAPI keyを呼び出す（情報漏洩リスクの回避）

```python
from google.colab import userdata
```

```python
os.environ["GOOGLE_API_KEY"] = userdata.get('APIkey')
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0.6)
```

プロンプトテンプレートの作成

```python
template = """
あなたは経験豊富なプロの調理師です。
ユーザーから提案された食材や料理について、プロならではの美味しく作るコツ（火加減や隠し味など）を交えながら、レシピを教えてください。
栄養面でのバランスの気遣いもお願いします。

お題: {recipe_query}
"""

prompt = PromptTemplate(
    input_variables=["recipe_query"],
    template=template,
)
```

ひとつの自動化された流れ（パイプライン）を作る

```python
chef_chain = prompt | llm | StrOutputParser()
```

botに質問してみる

```python
question = "ホットケーキを作りたいです。できればヘルシーに作りたい"
result = chef_chain.invoke({"recipe_query": question})

print("=== 調理師ボット(Gemini版)の回答 ===")
print(result)
```
